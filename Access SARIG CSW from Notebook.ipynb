# Access SARIG CSW from Notebook

## Notebook description

This notebook provides introductory examples for accessing the SARIG (South Australian Resource Information Gateway) catalogue through its OGC Catalogue Service for the Web (CSW) interface. It is intended as a starting point for inspecting service metadata and exploring catalogue-oriented workflows.

CSW is an OGC standard for discovering and querying metadata about geospatial resources. Common operations include:

- `GetCapabilities`: returns service metadata and supported operations
- `DescribeRecord`: describes supported metadata record structures
- `GetDomain`: returns domain values for a property
- `GetRecords`: searches for matching metadata records
- `GetRecordById`: returns a record by identifier
- `GetRepositoryItem`: returns a repository item when supported

## Possible Python use cases

### 1. Retrieve CSW capabilities


```python
import requests

url = "https://catalog.sarig.sa.gov.au/csw?service=CSW&request=GetCapabilities&version=2.0.2"
response = requests.get(url, timeout=30)
response.raise_for_status()

# CSW capabilities are returned as XML, not JSON.
print(response.text[:2000])
```

```python
getdomain_url = "https://catalog.sarig.sa.gov.au/csw?service=CSW&request=GetDomain&version=2.0.2&PropertyName=Title"
response = requests.get(getdomain_url, timeout=30)
response.raise_for_status()

# GetDomain also returns XML.
print(response.text[:2000])
```
